```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('Customer Churn.csv')
```

```python
data.head(5)
```

```python
data.info()
```

```python
# perform the operation on total charges as replace the blank with 0 and convert it into float as datatype
```

```python
data['TotalCharges']=data['TotalCharges'].replace(" ","0")
data['TotalCharges']=data['TotalCharges'].astype('float')
```

```python
data.info()
```

```python
data.isnull().sum().sum()
```

```python
data.describe()
```

```python
data['customerID'].duplicated().sum()
```

```python
#covert value of senior citizen for 0 with 'no' and 1 with 'yes'
```

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return "no"

data['SeniorCitizen']=data['SeniorCitizen'].apply(conv)
```

```python
data.head()
```

```python
ax=sns.countplot(x='Churn' ,data=data)
ax.bar_label(ax.containers[0])
plt.title("Count of Customer By Churn")
plt.show()
```

```python
gb=data.groupby("Churn").agg({'Churn':'count'})
gb
```

```python
plt.figure(figsize=(3,4))
plt.pie(gb['Churn'],labels=gb.index, autopct="%1.2f%%")
plt.title("Customer Of Churned Customers" , fontsize=10)
plt.show()
```

```python
#from the given pie chart we can conclude that 26.54% have chruned out
#now let's explore beh
```

```python
plt.figure(figsize=(4.5,4))
ax=sns.countplot(x="gender", data=data, hue="Churn")
ax.bar_label(ax.containers[0])
plt.title("Churn By Gender")
plt.show() 
```

```python
plt.figure(figsize=(4.5,4))
sns.countplot(x="SeniorCitizen", data=data, hue="Churn" , palette=["#A8D8EA" ,"#FFD6A5"])
plt.title("Churn By Senior Citizen")
plt.show() 
```

```python
#Display histrogram
plt.figure(figsize=(9,4))
sns.histplot(x="tenure" , data=data,  bins=72, hue="Churn")
plt.show()
```

```python
columns = ['PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 
           'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

# Number of columns for the subplot grid (you can change this)
n_cols = 3
n_rows = (len(columns) + n_cols - 1) // n_cols  # Calculate number of rows needed

# Create subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 4))  # Adjust figsize as needed

# Flatten the axes array for easy iteration (handles both 1D and 2D arrays)
axes = axes.flatten()

# Iterate over columns and plot count plots
for i, col in enumerate(columns):
    sns.countplot(x=col, data=data, ax=axes[i], hue = data["Churn"])
    axes[i].set_title(f'Count Plot of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')

# Remove empty subplots (if any)
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (6,4))
ax = sns.countplot(x = "PaymentMethod", data = data, hue = "Churn")
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.title("Churned Customers by Payment Method")
plt.xticks(rotation = 45)
plt.show()
```
